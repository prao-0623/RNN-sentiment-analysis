```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
#define vocabulary size
vocab_size = 1000
```

```python
# One hot represantation for every word
one_hot_repr = [one_hot(words,vocab_size)for words in sent]
one_hot_repr
```

```python
# word embedding repersentation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences #pad sequence used to make all sentence of equal size
from tensorflow.keras.models import Sequential
```

```python
import numpy as np


sent_length = 8
embeded_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(embeded_docs)
```

```python
# feature representation
dim = 10

```

```python
model = Sequential()
model.add(Embedding(vocab_size,dim,input_length=sent_length))
model.compile('adam','mse')
model.summary()
```

```python
model.predict(embeded_docs[0])

# array([  0,   0,   0,   0, 243, 317, 114, 176], dtype=int32) -- This will be converted to below output
#where 243 - will be converted to [ 0.00872958,  0.00793471,  0.04524699, -0.01686832, -0.04508636,
        # -0.01705282, -0.01197277,  0.01483649, -0.02599841, -0.01468022] -- 10 dim
```
